# Construct a cloud database by Render

```python
import psycopg2
from dotenv import load_dotenv
import os

load_dotenv()

def create_connection():
    """
    1. link to cloud postgres DB
    2. establish the environment parameter of the template
    """    
    try:
        conn = psycopg2.connect(
            dbname=os.getenv("DBNAME"),
            user=os.getenv("USER"),
            password=os.getenv("PASSWORD"),
            host=os.getenv("HOST"),
            port="5432"
        )
        return conn
    except psycopg2.Error as e:
        print(f"Error connecting to the database: {e}")
        return None


def get_taipei_station_count(cursor):
    query = """
    SELECT COUNT(*) FROM station_info;
    """
    cursor.execute(query)
    result = cursor.fetchall()
    return result

def get_station_names(cursor):
    query = """
    SELECT distinct "stationName" FROM station_info;
    """
    cursor.execute(query)
    result = cursor.fetchall()
    # station as list
    result_list = [station[0] for station in result]
    
    return result_list
```

```python
def main():
    conn = psycopg2.connect(
            dbname=os.getenv("DBNAME"),
            user=os.getenv("USER"),
            password=os.getenv("PASSWORD"),
            host=os.getenv("HOST"),
            port="5432"
        )
    
    if conn:
        print("Successfully connected to the database!")
        cursor = conn.cursor()
        # result = get_taipei_station_count(cursor=cursor)
        # print(result)
        result = get_station_names(cursor=cursor)
        print(result)
        cursor.close()
        conn.close()

    else:
        print("Failed to connect to the database.")


if __name__ == "__main__":
    main()
```
